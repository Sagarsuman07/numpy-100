#### 22. Normalize a 5x5 random matrix (★☆☆)

````python


#THIS SOLUTION IS FOR NORMALIZING VALUE BETWEEN 0 AND 1 CALLED Min-max normalization
# THERE ARE OTHER TYPES OF NORMALIZATION TOO ONE GIVEN IN ANSWER(22) IS Standardization 
#  WHICH NORMALIZE DATA AROUND 0



# Step 1: Create a random 5x5 matrix (values 0 to 1)
matrix = np.random.rand(5, 5)
print("Original matrix:\n", matrix)

# Step 2: Find min and max
min_val = matrix.min()
max_val = matrix.max()

# Step 3: Apply min-max normalization
normalized_matrix = (matrix - min_val) / (max_val - min_val)
print("\nNormalized matrix:\n", normalized_matrix)
````

#### 33. How to get the dates of yesterday, today and tomorrow? (★☆☆)

````python




#LEARN np.datetime64 and np.timedelta64



#DATE
today = np.datetime64('today', 'D')      
yesterday = today - np.timedelta64(1, 'D')
tomorrow = today + np.timedelta64(1, 'D')

print("Yesterday:", yesterday)
print("Today:", today)
print("Tomorrow:", tomorrow)


#DATE+TIME
now = np.datetime64('now')                
yesterday = now - np.timedelta64(1, 'D')
tomorrow = now + np.timedelta64(1, 'D')

print("Yesterday:", yesterday)
print("Now:", now)
print("Tomorrow:", tomorrow)
````

#### 42. Consider two random arrays A and B, check if they are equal (★★☆)

````python


# =========================================================
# Element-wise comparison
# =========================================================
# A == B performs element-wise comparison and returns
# a boolean array, NOT a single True/False value.

A = np.array([1, 2, 3])
B = np.array([1, 2, 4])

print("A == B :", A == B)     # [ True  True False ]


# =========================================================
# np.all()
# =========================================================
# np.all() returns True ONLY if all elements are True

print("np.all(A == B):", np.all(A == B))


# =========================================================
# Another use of np.all()
# Check if all elements satisfy a condition

A = np.array([2, 4, 6])
print("All elements even:", np.all(A % 2 == 0))


# =========================================================
# Floating-point comparison issue
# =========================================================
# Floating-point numbers are not exact due to binary storage.
# Example: 0.1 + 0.2 is NOT exactly equal to 0.3

a = 0.1 + 0.2
b = 0.3

print("a == b :", a == b)
print("np.allclose(a, b):", np.allclose(a, b))


# =========================================================
# np.array_equal()
# =========================================================
# np.array_equal() checks:
# - Same shape
# - Same values
# - Same dtype

A = np.array([1, 2, 3])
B = np.array([1, 2, 3])

print("np.array_equal(A, B):", np.array_equal(A, B))
````
